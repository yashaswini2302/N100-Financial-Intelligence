```python
import pandas as pd
import os

RAW_PATH = "../data/raw"

excel_files = [f for f in os.listdir(RAW_PATH) if f.endswith(".xlsx")]

datasets = {}

for file in excel_files:
    path = os.path.join(RAW_PATH, file)
    datasets[file] = pd.read_excel(path)

print("Datasets Loaded:", len(datasets))

for name in datasets:
    print(name)
```

```python
import pandas as pd
import os

RAW_PATH = "../data/raw"

excel_files = [f for f in os.listdir(RAW_PATH) if f.endswith(".xlsx")]

datasets = {}

for file in excel_files:
    path = os.path.join(RAW_PATH, file)
    datasets[file] = pd.read_excel(path)

print("Datasets Loaded:", len(datasets))

for name in datasets:
    print(name)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/db/nifty100.db")

tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

print("Tables in Database:")
print(tables)
```

```python
for table in tables["name"]:
    df = pd.read_sql(f"SELECT * FROM {table}", conn)

    print("=" * 50)
    print(f"Table : {table}")
    print(f"Rows  : {len(df)}")
    print(df.head())
```

```python
conn.close()

print("Database connection closed successfully.")
```
